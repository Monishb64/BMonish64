```python
import pandas as pd
import sqlite3

df = pd.read_csv("driver_standings.csv")
```

```python
print("Duplicate Rows:", df.duplicated().sum())

print("Missing Values:")

print(df.isnull().sum())
```

```python
df.to_excel("driver_standings.xlsx", index=False)
print("Excel file is exported successfully!")
```

```python
conn = sqlite3.connect("driver_standings.db")

df.to_sql("driver_standings", conn, if_exists="replace", index=False)

print("Database is created successfully!")
```

```python
query = "SELECT * FROM driver_standings"

result = pd.read_sql_query(query, conn)

result.head()
```

```python
df.to_csv("driver_standings_export.csv", index=False)
print("CSV file exported successfully!")
```

```python
df.to_excel("driver_standings_export.xlsx", index=False)
print("Excel file is exported successfully!")
```

```python
df.to_json("driver_standings_export.json", orient="records")
print("JSON file exported successfully!")
```
